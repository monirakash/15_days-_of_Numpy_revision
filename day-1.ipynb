```python
# using python list 
result=[]
for a in numbers:
    num=a*2
    result.append(num)
print(result)
```

## there are four main reasons
* 1. Python storess references to objects in memory. every integer is a separate python object
    while numpy stores raw data in a continuous block of memory.
 * 2. Numpy is written in C. Python loop is slow because it is an interpreted language. Numpy is written in C and C is a compiled language. 
*  3. Numpy uses vectorization which is a technique of performing operations on entire arrays instead of individual elements. 
* 4. SIMD optimization. Numpy uses SIMD optimization which is a technique of performing operations on multiple data points simultaneously.
